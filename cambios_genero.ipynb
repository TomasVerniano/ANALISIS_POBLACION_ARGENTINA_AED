IMPORTAR BIBLIOTECAS

```python
import pandas as pd
import matplotlib.pyplot as plt
```

GENERAR DATAFRAMES

```python
cambios_genero_año = pd.read_csv("DATA/cambio_genero_genero_anio_cambio_mayo_2024.csv")
cambios_genero_año.info()

cambios_genero_año = cambios_genero_año[['provincia_nombre', 'anio_cambio', 'genero', 'cantidad']]
cambios_genero_año.head()
```

```python
cambios_genero_edad = pd.read_csv("DATA/cambio_genero_provincia_genero_edad_cambio_julio_2026.csv")
cambios_genero_edad.info()

cambios_genero_edad = cambios_genero_edad[['provincia_nombre', 'genero', 'edad_actual_grandes_grupos', 'cantidad']]
cambios_genero_edad.head()
```

MODIFICAR Y GRAFICAR DATAFRAMES 

```python
cambios_genero_año_total = cambios_genero_año.groupby('anio_cambio')['cantidad'].sum()
cambios_genero_año_total.plot(kind='line', x='anio_cambio', xlabel='Año', y='cantidad', title='Cambios de genero por año', marker='o', grid=True)
```

```python
cambios_genero_sexo = cambios_genero_edad.groupby('genero')['cantidad'].sum()
cambios_genero_sexo.plot(kind='bar', xlabel='Sexo', title= 'Cantidad por sexo', grid=True)
```

```python
cambios_genero_edad_total = cambios_genero_edad.groupby('edad_actual_grandes_grupos')['cantidad'].sum()
cambios_genero_edad_total.plot(kind='barh', x='edad_actual_grandes_grupos', ylabel='Edad', y='cantidad', title='Cantidad por rango de edad', grid=True)
```

```python
cambios_genero_edad_f = cambios_genero_edad[(cambios_genero_edad['genero'] == 'F')]
cambios_genero_edad_f = cambios_genero_edad_f.groupby('edad_actual_grandes_grupos')['cantidad'].sum()
cambios_genero_edad_f.plot(kind='barh', title='Cantidad por rango de edad y genero femenino', ylabel='Grupo de edad', grid=True)
```

```python
cambios_genero_edad_m = cambios_genero_edad[(cambios_genero_edad['genero'] == 'M')]
cambios_genero_edad_m = cambios_genero_edad_m.groupby('edad_actual_grandes_grupos')['cantidad'].sum()
cambios_genero_edad_m.plot(kind='barh', title='Cantidad por rango de edad y genero masculino', ylabel='Grupo de edad', grid=True)
```

```python
cambios_genero_provincia = cambios_genero_año.groupby('provincia_nombre')['cantidad'].sum()
cambios_genero_provincia.plot(kind='barh', ylabel='Distritos', title='Cambios de genero por distrito', grid=True)
```

GENERAR NUEVOS DATOS

```python
cambios_genero_año_total.to_csv("DATA_PROCESADA/cambios_genero_año_total.csv")

cambios_genero_edad_total.to_csv("DATA_PROCESADA/cambios_genero_edad_total.csv")

cambios_genero_sexo.to_csv("DATA_PROCESADA/cambios_genero_sexo.csv")

cambios_genero_edad_m.to_csv("DATA_PROCESADA/cambios_genero_edad_m.csv")

cambios_genero_edad_f.to_csv("DATA_PROCESADA/cambios_genero_edad_f.csv")

cambios_genero_provincia.to_csv("DATA_PROCESADA/cambios_genero_provincia.csv")
```
