IMPORTAR BIBLIOTECAS

```python
import pandas as pd
import matplotlib.pyplot as plt
```

GENERAR DATAFRAMES

```python
nacimientos_madres = pd.read_csv("DATA/cantidad_de_nacimientos_edad_madre_por_anio_de_nacimiento_2012_2024.csv")
nacimientos_madres.info()

nacimientos_madres = nacimientos_madres[['anio', 'provincia_nombre', 'grupo_etario_madre', 'cantidad']]
nacimientos_madres.head()
```

```python
nacimientos = pd.read_csv("DATA/nacimientos_por_departamento_y_anio_2012_2024.csv")
nacimientos.info()

nacimientos = nacimientos[['anio', 'provincia_nombre', 'nacimientos']]
nacimientos.head()
```

```python
cambios_genero_año = pd.read_csv("DATA/cambio_genero_genero_anio_cambio_mayo_2024.csv")
cambios_genero_año.info()

cambios_genero_año = cambios_genero_año[['provincia_nombre', 'genero', 'anio_cambio', 'cantidad']]
cambios_genero_año.head()
```

```python
cambios_genero_edad = pd.read_csv("DATA/cambio_genero_provincia_genero_edad_cambio_julio_2026.csv")
cambios_genero_edad.info()

cambios_genero_edad = cambios_genero_edad[['provincia_nombre', 'genero', 'edad_actual_grandes_grupos', 'cantidad']]
cambios_genero_edad.head()
```

MODIFICAR Y GRAFICAR DATAFRAMES 

```python
cambios_genero_año_total = cambios_genero_año.groupby('anio_cambio')['cantidad'].sum()
cambios_genero_año_max = int(cambios_genero_año_total.max())
cambios_genero_año_total.plot(kind='line', x='anio_cambio', xlabel='Año', y='cantidad', title='Cambios de genero por año', marker='o', grid=True)
```

```python
cambios_genero_edad_total = cambios_genero_edad.groupby('edad_actual_grandes_grupos')['cantidad'].sum()
cambios_genero_edad_total.plot(kind='barh', x='edad_actual_grandes_grupos', ylabel='Edad', y='cantidad', title='Grupo de edad sobre cambios de genero', grid=True)
```

```python
madres_grupo_edad_total = nacimientos_madres.groupby(['grupo_etario_madre', 'anio'])['cantidad'].sum()
madres_grupo_edad_total.plot(kind='barh', ylabel='Año y grupo de edad', grid=True, figsize=(20,20))
```

```python
madres_total_año = nacimientos_madres.groupby('anio')['cantidad'].sum()
madres_total_año.plot(kind='line', grid=True, title='Cantidad de madres', xlabel='Año', marker='o')
```

```python
nacimientos_total_año = nacimientos.groupby('anio')['nacimientos'].sum()
nacimientos_total_año.plot(kind='line', grid=True, title='Cantidad de nacimientos', xlabel='Año', marker='o')
```

GENERAR NUEVOS DATOS

```python
cambios_genero_año_total.to_csv("DATA_PROCESADA/cambios_genero_año_total.csv")

cambios_genero_edad_total.to_csv("DATA_PROCESADA/cambios_genero_edad_total.csv")

madres_grupo_edad_total.to_csv("DATA_PROCESADA/madres_grupo_edad_total.csv")

madres_total_año.to_csv("DATA_PROCESADA/madres_total_año.csv")

nacimientos_total_año.to_csv("DATA_PROCESADA/nacimientos_total_año.csv")
```
