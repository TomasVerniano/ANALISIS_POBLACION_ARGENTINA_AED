IMPORTAR BIBLIOTECAS

```python
import pandas as pd
import matplotlib.pyplot as plt
```

GENERAR DATAFRAMES

```python
inversion_socio_economica = pd.read_csv("DATA/inversion_objetivosocioeconomico.csv", sep=';', encoding='latin1')
inversion_socio_economica.info()

inversion_socio_economica.head()
```

```python
inversion_investigacion = pd.read_csv("DATA/inversion_disciplina.csv", sep=';')
inversion_investigacion.info()

inversion_investigacion.head()
```

MODIFICAR Y GRAFICAR DATAFRAMES

```python
inversion_salud = inversion_socio_economica[inversion_socio_economica['OBJET_SOCEC_ID'] == 'Salud']
inversion_salud_año = inversion_salud.groupby('ANIO')['INV_ID_PESOS_CORR'].sum()

inversion_salud_año.plot(marker='o', grid=True, xticks=inversion_salud_año.index, rot=45, title='Inversion en salud', xlabel='Año', ylabel='Millones de pesos')
```

```python
inversion_investigacion_salud = inversion_investigacion[inversion_investigacion['DISCIPL_ID'] == 'Ciencias M�dicas']
inversion_investigacion_salud_año = inversion_investigacion_salud.groupby('ANIO')['INV_ID_PESOS_CORR'].sum()

inversion_investigacion_salud_año.plot(marker='o', grid=True, xticks=inversion_investigacion_salud_año.index, rot=45, title='Inversion en investigacion de la salud', xlabel='Año', ylabel='Millones de pesos')
```

GENERAR NUEVOS DATOS

```python
inversion_salud_año.to_csv("DATA_PROCESADA/inversion_salud_año.csv")

inversion_investigacion_salud_año.to_csv("DATA_PROCESADA/inversion_investigacion_salud_año.csv")
```
